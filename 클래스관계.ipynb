```python
# 상속 관계
```

```python
# 부모 클래스
class Animal:
    def describe(self):
        print('동물이다')
```

```python
# 자식 클래스
class Dog(Animal): # 부모 클래스를 자식 클래스의 소괄호에 집어넣으면 상속 관계 성립
    pass
```

```python
puppy=Dog()
```

```python
puppy.describe() # 부모의 함수를 상속받았으므로 사용 가능
```

```python
# 포함 관계
```

```python
class Engine:
    def start(self):
        print('Engine started')
```

```python
class Car:
    def __init__(self):
        self.engine=Engine()
    def drive(self):
        self.engine.start()
        print('Car is driving')
```

```python
my_car=Car()
```

```python
my_car.drive()
```

```python
# 사용 관계
```

```python
class Knife:
    def cut(self, food):
        return f'{food}을/를 자른다'
```

```python
class Chef:
    def cook(self, knife, ingredient):
        result=knife.cut(ingredient)
        return f'요리 중...{result}'
```

```python
kim_chef=Chef()
k_knife=Knife()
```

```python
kim_chef.cook(k_knife, '양파')
```
